```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Cargamos el dataset limpio de la capa Silver de la cobertura municipal (UPME)
df_icee = pd.read_csv('../data/silver/cobertura_ciudades_clean.csv')
display(df_icee.head())
```

```python

#Dimensiones del Dataset

print(f"Número de filas: {df_icee.shape[0]}")
print(f"Número de columnas: {df_icee.shape[1]}")
print(f"Dimensiones completas (Filas, Columnas): {df_icee.shape}")
```

```python
#Tipos de Dato

df_icee.info()
```

```python
#Nulos o Faltantes

nulos_por_columna = df_icee.isnull().sum()
print("Valores nulos por columna:")
print(nulos_por_columna)
```

```python
#ciudades y años abarca el dataset

print("Ciudades en el dataset:", df_icee['ciudad'].unique())
print("Años analizados:", df_icee['anio'].unique())
print("Departamentos:", df_icee['departamento'].unique())
```

```python
#Gráfica de evolución del ICEE

plt.figure(figsize=(10, 6))
sns.lineplot(data=df_icee, x='anio', y='icee', hue='ciudad', marker='o', linewidth=2.5)
plt.title('Evolución del Índice de Cobertura de Energía Eléctrica (ICEE) 2021-2024', fontsize=14, fontweight='bold')
plt.xlabel('Año', fontsize=12)
plt.ylabel('ICEE (%)', fontsize=12)
plt.legend(title='Ciudad')
plt.ylim(90, 101)
plt.show()
```
